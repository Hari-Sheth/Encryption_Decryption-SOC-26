# Frequency Distribution: Before vs After
Visualizes the initial (original) vs final (decrypted) frequency magnitude distributions produced by `train_model.ipynb`.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Load the frequency axis and the initial/final magnitude spectra
`base_frequencies.npy` comes from `create_dataset.ipynb`.
`true_mags.npy` / `decrypted_mags.npy` come from the inference step in `train_model.ipynb`.

```python
data_dir = "dataset_files"

base_frequencies = np.load(os.path.join(data_dir, "base_frequencies.npy"))
true_mags = np.load(os.path.join(data_dir, "true_mags.npy"))
decrypted_mags = np.load(os.path.join(data_dir, "decrypted_mags.npy"))

# Average magnitude at each frequency bin, across all chunks of the test audio
avg_true = true_mags.mean(axis=0)
avg_decrypted = decrypted_mags.mean(axis=0)
```

## 2. Side-by-side comparison

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(base_frequencies, avg_true, color="tab:blue")
axes[0].set_title("Initial Frequency Distribution (Original Audio)")
axes[0].set_xlabel("Frequency (Hz)")
axes[0].set_ylabel("Average Magnitude")
axes[0].grid(alpha=0.3)

axes[1].plot(base_frequencies, avg_decrypted, color="tab:orange")
axes[1].set_title("Final Frequency Distribution (Decrypted Audio)")
axes[1].set_xlabel("Frequency (Hz)")
axes[1].set_ylabel("Average Magnitude")
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig("frequency_distribution_side_by_side.png", dpi=150)
plt.show()
```

## 3. Overlay comparison
Makes the before/after difference easy to see on a single plot.

```python
plt.figure(figsize=(10, 5))
plt.plot(base_frequencies, avg_true, label="Initial (Original)", color="tab:blue", alpha=0.8)
plt.plot(base_frequencies, avg_decrypted, label="Final (Decrypted)", color="tab:orange", alpha=0.8)
plt.title("Frequency Distribution: Original vs Decrypted")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Average Magnitude")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("frequency_distribution_overlay.png", dpi=150)
plt.show()

print("Saved plots: frequency_distribution_side_by_side.png, frequency_distribution_overlay.png")
```
